## 7. Create a Copy of the Dataset

```python
cleaned_df = df.copy()

print("Copy created successfully!")
```

## 8. Remove Duplicate Rows

```python
before = len(cleaned_df)

cleaned_df = cleaned_df.drop_duplicates()

after = len(cleaned_df)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)
```

## 9. Handle Missing Values in Age

```python
age_median = cleaned_df["Age"].median()

print("Median Age:", age_median)

cleaned_df["Age"] = cleaned_df["Age"].fillna(age_median)

print("Missing Age after cleaning:",
      cleaned_df["Age"].isnull().sum())
```

### 10. Handle Missing Values in Embarked

```python
embarked_mode = cleaned_df["Embarked"].mode()[0]

print("Mode of Embarked:", embarked_mode)

cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(embarked_mode)

print("Missing Embarked after cleaning:",
      cleaned_df["Embarked"].isnull().sum())
```

## 11. Standardize Text Values

```python
cleaned_df["Sex"] = cleaned_df["Sex"].str.strip().str.lower()
cleaned_df["Embarked"] = cleaned_df["Embarked"].str.strip().str.upper()

print("Text values standardized successfully!")
```

## 12. Remove Cabin Column

```python
cleaned_df = cleaned_df.drop(columns=["Cabin"])

print("Cabin column removed.")
print("New shape:", cleaned_df.shape)
```

## 13. Final Data Quality Validation

```python
print("FINAL DATA QUALITY CHECK")
print("------------------------")

print("\nMissing values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows:",
      cleaned_df.duplicated().sum())

print("\nDataset shape:",
      cleaned_df.shape)
```

## 14. Export the Cleaned Dataset

```python
cleaned_df.head()
```

```python
cleaned_df.to_csv("cleaned_titanic.csv", index=False)

print("Cleaned dataset saved successfully!")
```
